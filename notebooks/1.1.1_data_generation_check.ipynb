# 1.1.1 Data Generation Check

```python
import sys
import os
import numpy as np

project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.append(project_root)

from src.noise import AbsoluteGaussianNoise, AdaptiveGaussianNoise
from src.generators import Math2DGenerator
from src.visualization import plot_data_distribution
```

### 1. Funtzio Lineala

```python
print("Funtzio lineala")
m = 2.0
n = 5.0
linear_func = lambda x: m * x + n 
generator = Math2DGenerator(linear_func)
df = generator.generate(
    n_samples=200, 
    x_range=(0, 20), 
    noise_strategy=AdaptiveGaussianNoise(noise_ratio=0.1)
)

print(f"Datuak sortuta. Shape: {df.shape}")
print(df.head())

plot_data_distribution(df, title="Froga: Zaratadun Funtzio Lineala (m=2, n=5)")
```

### 2. Karratu Funtzioa

```python
print("Karratu funtzioa")
a = 0.7
b = 2.3
c = -2.0
sqr_func = lambda x: a*x*x + b*x +c 
generator = Math2DGenerator(sqr_func)
df = generator.generate(
    n_samples=200, 
    x_range=(0, 20),
    noise_strategy=AdaptiveGaussianNoise(noise_ratio=0.1)
)

print(f"Datuak sortuta. Shape: {df.shape}")
print(df.head())

plot_data_distribution(df, title="Zaratadun Karratu Funtzioa (a=0.7, b=2.3, c=-2.0)")
```

### 3. Funtzio Sinusoidala

```python
print("Funtzio sinusoidala")
A = 2.0
f = 1.2
phi = 0.72

sin_func = lambda x: A * np.sin(f*x + phi)
generator = Math2DGenerator(function=sin_func)
df = generator.generate(
        n_samples=200,
        x_range=(0, 20),
        noise_strategy=AdaptiveGaussianNoise(noise_ratio=0.1)
)

print(f"Datuak sortuta. Shape: {df.shape}")
print(df.head())

plot_data_distribution(df, title="Zaratadun Funtzio sinusoidala (A=2.0, f=1.2, phi=0.72)")
```

### 4. Borobil Arku Funtzioa

```python
print("Borobil arku funtzioa")
x_0 = 2.0
y_0 = 1
r = 5
circ_arch_func = lambda x: np.sqrt(np.maximum(0, r**2 - (x-x_0)**2)) + y_0
generator = Math2DGenerator(function=circ_arch_func)
df = generator.generate(
    n_samples=200,
    x_range=(0, 20),
    noise_strategy=AdaptiveGaussianNoise(noise_ratio=0.1)
)

print(f"Datuak sortuta. Shape: {df.shape}")
print(df.head())

plot_data_distribution(df, title="Froga: Zaratadun borobil arkua (r=5, x_0=2.0, y_0=-1.0)")
```
