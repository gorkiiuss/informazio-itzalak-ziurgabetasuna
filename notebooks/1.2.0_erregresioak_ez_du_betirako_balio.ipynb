# 1.2.0 Underfitting-a
**Helburua:** Erakutsi zer gertatzen den eredu sinpleegi bat (Erregresio Lineala) erabiltzen denean datu konplexuagoekin (Sinusoidala).

Aurreko esperimentuan ikusi da eredu lineala perfektua dela datu linealetarako. Baina mundu erreala ez da beti lerro zuzen bat.
Hemen ikusiko dugu **"Underfitting"**-aren fenomenoa: ereduak ez du nahikoa gaitasun datuen forma ulertzeko.

```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt

project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.append(project_root)

from src.generators import Math2DGenerator
from src.noise import AdaptiveGaussianNoise
from src.shadows import CircularShadow
from src.splitter import GapSplitter
from src.models import LinearRegressionModel
from src.orchestrator import RegressionOrchestator
from src.visualization import plot_experiment_results
```

### 1. Datu Ez-Linealak Sortu (Sinusoidala)

```python
sine_func = lambda x: 5.0 * np.sin(0.5 * x) + 10.0
generator = Math2DGenerator(sine_func, seed=42)
noise = AdaptiveGaussianNoise(noise_ratio=0.1, seed=42)
```

### 2. Itzala Definitu
Zirkulu bat jarriko dugu "olatua"ren beheko partean (haran batean).

```python
shadow = CircularShadow(x_0=10, y_0=5, r=4.0)
```

### 3. Exekutatu (Eredu Linealarekin)

```python
splitter = GapSplitter(test_size=0.2, random_state=42)
model = LinearRegressionModel() 

orchestrator = RegressionOrchestator(generator, splitter, model)

results = orchestrator.run(
    n_samples=400,
    x_range=(0, 20),
    noise_strategy=noise,
    shadow=shadow
)
```

### 4. Emaitzak eta Ondorioak

```python
metrics = results.metrics
print("--- EMAITZAK (Linear Model vs Sine Data) ---")
print(f"Train MSE:      {metrics.mse_train:.4f} (Altua izango da)")
print(f"Shadow Test MSE:{metrics.mse_test_shadow:.4f} (Oso altua)")

plot_experiment_results(
    results.split_data,
    results.predictions,
    title="FAILURE CASE: Linear Regression on Sine Wave"
)
```
