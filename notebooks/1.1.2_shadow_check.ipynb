# 1.1.2 Shadow Check

```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt 

project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.append(project_root)

from src.generators import Math2DGenerator
from src.noise import AbsoluteGaussianNoise
from src.shadows import CircularShadow, BandShadow
from src.visualization import plot_shadow_distribution
```

### Datuak

```python
print("--- Itzalen Froga Hasten ---")

linear_func = lambda x: x 
generator = Math2DGenerator(linear_func, seed=42)

df = generator.generate(
    n_samples=500, 
    x_range=(0, 20), 
    noise_strategy=AbsoluteGaussianNoise(std_dev=0.5)
)
print(f"Sortutako datuak: {df.shape}")
```

### 1. Froga: Borobil Itzala

```python
print("\n1. Froga: Borobil Itzala")
shadow_circ = CircularShadow(x_0=10, y_0=10, r=3.0)
mask_circ = shadow_circ.is_in_shadow(df['x'].values, df['y'].values)

n_shadow = np.sum(mask_circ)
print(f"Puntuak guztira: {len(df)}")
print(f"Itzalpean (Franja): {n_shadow} ({n_shadow/len(df)*100:.1f}%)")

plot_shadow_distribution(
    df, mask_circ, 
    title="1. Froga: Borobil Itzala (x_0=10, y_0=10, r=3)"
)
```

### 2. Froga: X ardatzan hutsunea

```python
print("\n2. Froga: X ardatzan hutsunea")
shadow_band = BandShadow(axis='x', min_val=10.0, max_val=12.0)
mask_band = shadow_band.is_in_shadow(df['x'].values, df['y'].values)

n_shadow = np.sum(mask_band)
print(f"Puntuak guztira: {len(df)}")
print(f"Itzalpean (Franja): {n_shadow} ({n_shadow/len(df)*100:.1f}%)")

plot_shadow_distribution(
    df, mask_band, 
    title="2. Froga: X ardatzan hutsunea(min_val=10.0, max_val=12.0)"
)
```
